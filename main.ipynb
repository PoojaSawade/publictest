```python
!pip install -r requirements.txt
from warpdrive import WarpDrive


wd = WarpDrive()

# your code here
from catboost import CatBoostClassifier
import pandas as pd

df = pd.DataFrame({
    "age": [25, 30, 22, 40],
    "income": [50000, 60000, 45000, 80000],
    "Gender": [0, 1, 0, 1],  # Target column
})

exog_columns = ["age", "income"]
col = exog_columns

clf = CatBoostClassifier(verbose=0)
clf.fit(df[exog_columns], df["Gender"])

wd.create_model(
    model=clf,
    library="catboost",
    model_technique="CatboostClassifier",
    input_variables=col,
    target_column="Gender",
    train_table="",
    lags=0,
    exog_columns=exog_columns,
)

```
